## 1. Can Python Understand Language?

Before running the code, look at these two sentences:

- "I love this movie."
- "I do not love this movie."

### Think first
1. Which sentence is positive?
2. Which sentence is negative?
3. What word appears in both sentences?
4. If a program only searched for the word `love`, would that be enough to understand the meaning?

```python
text1 = "I love this movie."
text2 = "I do not love this movie."

print("love" in text1)
print("love" in text2)
```

## 3. Same word, different meaning
Consider:

- "Apple released a new laptop."
- "She ate an apple."

### Think first
Does the word `apple` mean the same thing in both sentences?

```python
text1 = "Apple released a new laptop."
text2 = "She ate an apple."

print("apple" in text1.lower())
print("apple" in text2.lower())
```

Python can detect the same sequence of characters.

But can it determine what `apple` refers to?

This is an example of why **context** matters in NLP.

## 4. Small changes, big changes in meaning

## 5. What can simple string operations tell us?

We can build a simple rule detecting the word `amazing` in the reviews.

But dies it correctly determine sentiment in every example?

What would a more intelligent NLP system need to consider besides individual words?

```python
reviews = [
    "Amazing product! 😍",
    "Not amazing at all.",
    "The product arrived yesterday.",
    "AMAZING!!!",
]

for review in reviews:
    print(review, "->", "amazing" in review.lower())
```
