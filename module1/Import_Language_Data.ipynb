### Extract all text for processing

Option 1: keep adding processed text to the string

Option 2: Create an array - every page is a separate element. Then join them

```python
# Option 1
pdf_data = ""
number_of_pages = len(reader.pages)
for page_number in range(number_of_pages):   # use xrange in Py2
    page = reader.pages[page_number]
    page_content = page.extract_text()
    pdf_data+=page_content
```

```python
# Option 2
pdf_data = []
number_of_pages = len(reader.pages)
for page_number in range(number_of_pages):   # use xrange in Py2
    page = reader.pages[page_number]
    page_content = page.extract_text()
    pdf_data.append(page_content)
' '.join(pdf_data)
```

## HTML Import

```python
from bs4 import BeautifulSoup
```

```python
import requests

# Define a User-Agent header to mimic a browser
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'}
url = 'https://en.wikipedia.org/wiki/Natural_language_processing'

# Make the GET request with the defined headers
response = requests.get(url, headers=headers)

# Raise an HTTPError for bad responses (4xx or 5xx status codes)
response.raise_for_status()

# Get the content as text
html_doc = response.text
```

```python
soup = BeautifulSoup(html_doc, 'html.parser')
strhtm = soup.prettify()
print(strhtm[:1000])
```

```python
print(soup.title)
print(soup.title.string)
#for x in soup.find_all('p'): print(x.text)
```
